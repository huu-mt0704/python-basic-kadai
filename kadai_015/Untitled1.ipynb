```python
class Human:
    # 1. コンストラクタ（名前と年齢を登録する）
    def __init__(self, name, age):
        self.name = name
        self.age = age

    # 2. 名前と年齢を print するメソッド（printinfo）
    def printinfo(self):
        print(self.name)
        print(self.age)


# 3. Humanクラスから実物（インスタンス）を1人作って変数に入れる
human = Human("侍太郎", 36)

# 4. 作ったメソッドを呼び出して画面に出力する
human.printinfo()
```
